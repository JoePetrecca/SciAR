## Paired test-sampling uncertainty
Components connect pairs sharing first-author identities, transitively retaining repeated-paper dependence. Each replicate uses identical component multiplicities for both models and all fixed seeds. Intervals describe the mean paired per-seed AUC difference conditional on these seeds. They do not include uncertainty over a population of training seeds. Few or giant components can make inference unreliable or impossible. No multiplicity correction or significance declaration is made.

```python
bootstrap = json.loads((RESULTS / 'bootstrap.json').read_text())
print(bootstrap['method'])
print(bootstrap['caveats'])
display(pd.DataFrame(bootstrap['comparisons']).T)
```
